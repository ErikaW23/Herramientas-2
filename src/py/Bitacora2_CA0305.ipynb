# Ejercicio 1: capa lineal hacia adelante <br>

Implemente la función linear_forward(X, W, b), la cual  implemente el paso hacia adelante de una capa lineal. 

**Operación matemática:** Y = XW + b, donde: 

- X es la matriz de entrada de tamaño (N, D(in)) <br>
- W es la matriz de pesos de tamaño (D(in), D(out)) <br>
- b es el vector de sesgo de la forma (D(out), ) <br>
- Y es la matriz resultante de tamaño (N, D(out)) <br>

*Requerimientos* 

1. Use la multiplicación matricial (@ or np.dot) para XW. <br>
2. Use broadcasting para la suma del vector del sesgo. <br>
3. Retorne un resultado de tamaño (N, D(out)). <br>
4. Maneje los lotes de forma eficiente: de manera vectorizada, sin usar ciclos. <br>
5. Asegúrese de que las dimensiones sean compatibles: D(in) debe coincidir entre X y W.
